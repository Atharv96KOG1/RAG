```python
import os
import requests

PDF_URL = "https://www.iima.ac.in/sites/default/files/2022-12/HR%20Policy%20Manual%202023.pdf"

PDF_PATH = "hr_policy_manual_2023.pdf"

response = requests.get(PDF_URL, timeout=60)
response.raise_for_status()

with open(PDF_PATH, "wb") as f:
    f.write(response.content)

print(f"PDF downloaded: {PDF_PATH}")

pipeline_options = PdfPipelineOptions()

# OCR for scanned/image-based content
pipeline_options.do_ocr = True

# Extract tables
pipeline_options.do_table_structure = True

# Generate page images
pipeline_options.generate_page_images = True

# Generate individual pictures/images
pipeline_options.generate_picture_images = True


from docling.document_converter import PdfFormatOption

doc_converter = DocumentConverter(
    format_options={
        "pdf": PdfFormatOption(
            pipeline_options=pipeline_options
        )
    }
)


print("Processing PDF...")

result = doc_converter.convert(PDF_PATH)

document = result.document

print("PDF processing completed!")



OUTPUT_DIR = "hr_policy_output"

os.makedirs(OUTPUT_DIR, exist_ok=True)



markdown = document.export_to_markdown(
    image_mode="referenced"
)

markdown_path = os.path.join(
    OUTPUT_DIR,
    "hr_policy_manual.md"
)

with open(markdown_path, "w", encoding="utf-8") as f:
    f.write(markdown)

print(f"Markdown saved: {markdown_path}")



json_path = os.path.join(
    OUTPUT_DIR,
    "hr_policy_manual.json"
)

document.save_as_json(json_path)

print(f"JSON saved: {json_path}")



IMAGE_DIR = os.path.join(
    OUTPUT_DIR,
    "images"
)

os.makedirs(IMAGE_DIR, exist_ok=True)

picture_count = 0

for item, level in document.iterate_items():

    # Check whether item contains an image
    if hasattr(item, "get_image"):

        try:
            image = item.get_image(document)

            if image is not None:

                image_path = os.path.join(
                    IMAGE_DIR,
                    f"image_{picture_count}.png"
                )

                image.save(image_path)

                picture_count += 1

        except Exception as e:
            print(
                f"Could not extract image: {e}"
            )


print(f"Images extracted: {picture_count}")



print("\n" + "=" * 80)
print("EXTRACTED CONTENT")
print("=" * 80)

print(markdown)

print("\nDone!")
```

```python
import json
import re
from pathlib import Path

from docling.document_converter import DocumentConverter
from docling.datamodel.pipeline_options import PdfPipelineOptions
from docling.document_converter import PdfFormatOption
from docling.datamodel.base_models import InputFormat

PDF_PATH = "hr_policy_manual_2023.pdf"



pipeline_options = PdfPipelineOptions()

# OCR
pipeline_options.do_ocr = True

# Table extraction
pipeline_options.do_table_structure = True

# Images
pipeline_options.generate_page_images = True
pipeline_options.generate_picture_images = True



converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(
            pipeline_options=pipeline_options
        )
    }
)



result = converter.convert(PDF_PATH)

document = result.document



def clean_text(text: str) -> str:

    if not text:
        return ""

    # Remove HTML tags
    text = re.sub(
        r"<[^>]+>",
        "",
        text
    )

    # Remove HTML comments
    text = re.sub(
        r"<!--.*?-->",
        "",
        text,
        flags=re.DOTALL
    )

    # Remove control characters
    text = re.sub(
        r"[\x00-\x08\x0B\x0C\x0E-\x1F\x7F]",
        "",
        text
    )

    # Normalize whitespace
    text = re.sub(
        r"[ \t]+",
        " ",
        text
    )

    text = re.sub(
        r"\n{3,}",
        "\n\n",
        text
    )

    return text.strip()




output = {
    "document": {
        "filename": Path(PDF_PATH).name,
        "content": []
    }
}



for item, level in document.iterate_items():

    if item.__class__.__name__ in [
        "TextItem",
        "SectionHeaderItem"
    ]:

        text = clean_text(
            item.text
        )

        if text:

            output["document"]["content"].append({
                "type": "text",
                "content": text
            })


    # --------------------------------------------------------
    # TABLE
    # --------------------------------------------------------

    elif item.__class__.__name__ == "TableItem":

        table_data = []

        try:

            # Convert Docling table → dataframe
            df = item.export_to_dataframe()

            # Clean column names
            df.columns = [
                clean_text(str(col))
                for col in df.columns
            ]

            # Convert rows to dictionaries
            for row in df.to_dict(
                orient="records"
            ):

                cleaned_row = {}

                for key, value in row.items():

                    cleaned_row[
                        clean_text(str(key))
                    ] = clean_text(
                        str(value)
                    )

                table_data.append(
                    cleaned_row
                )

        except Exception as e:

            print(
                f"Table extraction error: {e}"
            )

        output["document"]["content"].append({
            "type": "table",
            "data": table_data
        })


# ============================================================
# 8. Save JSON
# ============================================================

OUTPUT_PATH = "hr_policy_structured.json"

with open(
    OUTPUT_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        output,
        f,
        ensure_ascii=False,
        indent=2
    )


print("Done!")
print(
    f"JSON saved to: {OUTPUT_PATH}"
)
```
